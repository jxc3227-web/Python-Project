```python
import random
import pandas as pd
```

```python
def create_deck():

    deck = [2, 3, 4, 5, 6, 7, 8, 9, 10,
            10, 10, 10, 11] * 4

    random.shuffle(deck)

    return deck
```

```python
def deal_card(deck):

    return deck.pop()
```

```python
def calculate_hand(hand):

    total = sum(hand)

    while total > 21 and 11 in hand:

        hand[hand.index(11)] = 1
        total = sum(hand)

    return total
```

```python
class Player:

    def __init__(self, name):

        self.name = name
        self.hand = []

    def receive_card(self, card):

        self.hand.append(card)

    def show_hand(self):

        print(self.name, ":", self.hand)
        print("Score:", calculate_hand(self.hand))
```

```python
def player_turn(deck, player):

    while calculate_hand(player.hand) < 21:

        choice = input("Hit or Stay? ").lower()

        if choice == "hit":

            player.receive_card(deal_card(deck))
            player.show_hand()

        elif choice == "stay":

            break

        else:

            print("Invalid choice!")

    if calculate_hand(player.hand) > 21:

        print("You busted!")
```

```python
def dealer_turn(deck, dealer):

    while calculate_hand(dealer.hand) < 17:

        dealer.receive_card(deal_card(deck))

    dealer.show_hand()
```

```python
def determine_winner(player, dealer):

    player_score = calculate_hand(player.hand)
    dealer_score = calculate_hand(dealer.hand)

    if player_score > 21:

        print("You lose!")

    elif dealer_score > 21:

        print("You win!")

    elif player_score > dealer_score:

        print("You win!")

    elif player_score < dealer_score:

        print("Dealer wins!")

    else:

        print("It's a tie!")
```

```python
def play_blackjack():

    deck = create_deck()

    player = Player("Player")
    dealer = Player("Dealer")

    for i in range(2):

        player.receive_card(deal_card(deck))
        dealer.receive_card(deal_card(deck))

    print("Blackjack")

    player.show_hand()

    print("Dealer's visible card:", dealer.hand[0])

    player_turn(deck, player)

    if calculate_hand(player.hand) <= 21:

        dealer_turn(deck, dealer)

    determine_winner(player, dealer)

    return player.hand, dealer.hand
```

```python
game_history = []

while True:

    player_hand, dealer_hand = play_blackjack()

    player_score = calculate_hand(player_hand)
    dealer_score = calculate_hand(dealer_hand)

    if player_score > 21:

        result = "Loss"

    elif dealer_score > 21 or player_score > dealer_score:

        result = "Win"

    elif player_score < dealer_score:

        result = "Loss"

    else:

        result = "Tie"

    game_history.append({
        "Player Score": player_score,
        "Dealer Score": dealer_score,
        "Result": result
    })

    again = input("\nPlay again? (yes/no): ").lower()

    if again != "yes":

        break
```
